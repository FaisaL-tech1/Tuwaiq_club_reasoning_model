```python
# Print versions in a way that doesn't depend on package distribution metadata.
# (Notebook scratch cells often run from source, so `importlib.metadata.version()` can fail.)

modules = {
    "reasoning_from_scratch": "reasoning_from_scratch",
    "torch": "torch",
    "tokenizers": "tokenizers",
}

for label, module_name in modules.items():
    try:
        module = __import__(module_name)
        mod_version = getattr(module, "__version__", "unknown")
        print(f"{label} version: {mod_version}")
    except ModuleNotFoundError:
        print(f"{label} version: not installed")
```
